# **1. IMPORT LIBRARY**

```python
import os
import pandas as pd
from pathlib import Path
```

# **2. MAPPING LABEL**

```python
DATASET_PATH = Path("../data/raw")
LABEL_MAPPING = {
    "genuine_accounts": "human",
    "social_spambots_1": "bot",
    "social_spambots_2": "bot",
    "social_spambots_3": "bot",
    "traditional_spambots_1": "bot",
    "traditional_spambots_2": "bot",
    "traditional_spambots_3": "bot",
    "traditional_spambots_4": "bot",
    "fake_followers": "bot"
}
```

# **3. LOAD USER**

```python
user_frames = []
for folder_name, label in LABEL_MAPPING.items():
    user_file = DATASET_PATH / f"{folder_name}.csv" / "users.csv"

    if not user_file.exists():
        print(f"users.csv tidak ditemukan : {folder_name}")
        continue

    df = pd.read_csv(user_file)
    df["label"] = label
    df["source_folder"] = folder_name
    user_frames.append(df)

users_all = pd.concat(user_frames, ignore_index=True)
print(users_all.shape)
users_all.head()
```

# **4. SAVE USERS**

```python
OUTPUT_DIR = Path("../data/interim")

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

users_all.to_csv(
    OUTPUT_DIR / "users_all.csv",
    index=False
)
```

# **5. LOAD TWEETS**

```python
tweet_frames = []
for folder_name, label in LABEL_MAPPING.items():
    tweet_file = DATASET_PATH / f"{folder_name}.csv" / "tweets.csv"

    if not tweet_file.exists():
        print(f"tweets.csv tidak ditemukan : {folder_name}")
        continue

    try:
        df = pd.read_csv(tweet_file, encoding="ISO-8859-1")
    except UnicodeDecodeError:
        df = pd.read_csv(tweet_file, encoding="utf-8", encoding_errors="replace")

    df["label"] = label
    df["source_folder"] = folder_name
    tweet_frames.append(df)

tweets_all = pd.concat(tweet_frames, ignore_index=True)
print(tweets_all.shape)
tweets_all.head()

tweets_all.to_csv(
    OUTPUT_DIR / "tweets_all.csv",
    index=False
)
```
