# **1. IMPORT LIBRARY**

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

# **2. LOAD & CHECK**

```python
DATA_PATH = Path("../data/interim")

users = pd.read_csv(DATA_PATH / "users_all.csv")
tweets = pd.read_csv(DATA_PATH / "tweets_all.csv")

print(users.shape)
users.head()

print(tweets.shape)
tweets.head()
```

# **3. CEK MISSING VALUE**

```python
users.isnull().sum().sort_values(ascending=False)

tweets.isnull().sum().sort_values(ascending=False)

#DROP DUPLICATE
users = users.drop_duplicates()

tweets = tweets.drop_duplicates()
```

# **6. CONVERT DATETIME**

```python
users["created_at"] = pd.to_datetime(
    users["created_at"],
    errors="coerce"
)

tweets["created_at"] = pd.to_datetime(
    tweets["created_at"],
    errors="coerce"
)
```

```python
users["timestamp_fallback"] = pd.to_datetime(users["timestamp"], errors="coerce")
mask_nat = users["created_at"].isna()
users.loc[mask_nat, "created_at"] = users.loc[mask_nat, "timestamp_fallback"]
users = users.drop(columns=["timestamp_fallback"])

print("NaT tersisa:", users["created_at"].isna().sum())

tweets["timestamp_fallback"] = pd.to_datetime(tweets["timestamp"], errors="coerce")
mask_nat = tweets["created_at"].isna()
tweets.loc[mask_nat, "created_at"] = tweets.loc[mask_nat, "timestamp_fallback"]
tweets = tweets.drop(columns=["timestamp_fallback"])

print("NaT tersisa:", tweets["created_at"].isna().sum())
```

# **7. CEK MISS AGAIN**

```python
users.isnull().sum()
# users = users.dropna()

numeric_cols = users.select_dtypes(include=np.number).columns

users[numeric_cols] = users[numeric_cols].fillna(0)

object_cols = users.select_dtypes(include="object").columns

users[object_cols] = users[object_cols].fillna("")
```

# **8. BOOLEAN**

```python
boolean_cols = [
    "verified",
    "geo_enabled",
    "protected"
]

for col in boolean_cols:
    if col in users.columns:
        users[col] = users[col].astype(int)
```

# **9. DROP COLUMN**

```python
drop_cols = [

    "profile_banner_url",

    "profile_image_url",

    "url"

]

users = users.drop(
    columns=drop_cols,
    errors="ignore"
)

users.info()
tweets.info()
```

# **10. SAVE**

```python
OUTPUT = Path("../data/processed")

OUTPUT.mkdir(
    parents=True,
    exist_ok=True
)

users.to_csv(
    OUTPUT / "users_clean.csv",
    index=False
)

tweets.to_csv(
    OUTPUT / "tweets_clean.csv",
    index=False
)
```
