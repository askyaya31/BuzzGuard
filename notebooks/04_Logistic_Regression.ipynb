# 1. IMPORT LIBRARY

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
```

# 2. LOAD DATASET

```python
DATA_PATH = Path("../data/processed")

feature_A = pd.read_csv(DATA_PATH / "feature_A.csv")
feature_AB = pd.read_csv(DATA_PATH / "feature_AB.csv")

feature_A = feature_A.fillna(0)
feature_AB = feature_AB.fillna(0)

feature_A.head()
feature_AB.head()
```

# 3. FEATURE A

split feature & label

```python
X_A = feature_A.drop(columns=["user_id", "label", "source_folder"])

y_A = feature_A["label"]
```

Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_A,
    y_A,
    test_size=0.2,
    random_state=42,
    stratify=y_A
)
```

Standard Scaler

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

Training

```python
lr_A = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_A.fit(X_train, y_train)
```

Prediction

```python
y_pred = lr_A.predict(X_test)
```

Evaluation

```python
acc_A = accuracy_score(y_test, y_pred)

pre_A = precision_score(
    y_test,
    y_pred,
    average="macro"
)

rec_A = recall_score(
    y_test,
    y_pred,
    average="macro"
)

f1_A = f1_score(
    y_test,
    y_pred,
    average="macro"
)

print("Accuracy :", acc_A)
print("Precision:", pre_A)
print("Recall   :", rec_A)
print("F1 Score :", f1_A)
```

Classification Report

```python
print(classification_report(
    y_test,
    y_pred
))
```

# 4. FEATURE A+B

split feature

```python
X_A = feature_A.drop(columns=["user_id", "label", "source_folder"], errors="ignore")
X_AB = feature_AB.drop(columns=["user_id", "label", "source_folder"], errors="ignore")
y_AB = feature_AB["label"]
```

Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_AB,
    y_AB,
    test_size=0.2,
    random_state=42,
    stratify=y_AB
)
```

Standard Scaler

```python
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

Training

```python
lr_AB = LogisticRegression(
    max_iter=1000,
    random_state=42
)

lr_AB.fit(X_train, y_train)
```

Prediction

```python
y_pred = lr_AB.predict(X_test)
```

Evaluation

```python
acc_AB = accuracy_score(y_test, y_pred)

pre_AB = precision_score(
    y_test,
    y_pred,
    average="macro"
)

rec_AB = recall_score(
    y_test,
    y_pred,
    average="macro"
)

f1_AB = f1_score(
    y_test,
    y_pred,
    average="macro"
)

print("Accuracy :", acc_AB)
print("Precision:", pre_AB)
print("Recall   :", rec_AB)
print("F1 Score :", f1_AB)
```

Classification Report

```python
print(classification_report(
    y_test,
    y_pred
))
```

# 5. FINAL RESULT

```python
results = pd.DataFrame({
    "Feature": ["A", "A+B"],
    "Accuracy": [acc_A, acc_AB],
    "Precision": [pre_A, pre_AB],
    "Recall": [rec_A, rec_AB],
    "F1 Score": [f1_A, f1_AB]
})

results
```

```python
RESULT_PATH = Path("../results")
RESULT_PATH.mkdir(parents=True, exist_ok=True)
results.to_csv(RESULT_PATH / "logistic_regression_results.csv", index=False)
```
