# 1. IMPORT LIBRARY

```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
```

# 2. LOAD RESULT

```python
RESULT_PATH = Path("../results")

lr = pd.read_csv(
    RESULT_PATH / "logistic_regression_results.csv"
)

rf = pd.read_csv(
    RESULT_PATH / "random_forest_results.csv"
)

xgb = pd.read_csv(
    RESULT_PATH / "xgboost_results.csv"
)
```

# 3. MERGE

```python
lr["Model"] = "Logistic Regression"

rf["Model"] = "Random Forest"

xgb["Model"] = "XGBoost"

comparison = pd.concat([
    lr,
    rf,
    xgb
], ignore_index=True)

comparison
```

# 4. SAVE COMPARE

```python
comparison.to_csv(
    RESULT_PATH / "model_comparison.csv",
    index=False
)
```

# 5. ACCURACY COMPARISON

```python
plt.figure(figsize=(10,5))

sns.barplot(

    data=comparison,

    x="Model",

    y="Accuracy",

    hue="Feature"

)

plt.title("Accuracy Comparison")

plt.ylim(0,1)

plt.show()
```

# 6. PRECISION

```python
plt.figure(figsize=(10,5))

sns.barplot(

    data=comparison,

    x="Model",

    y="Precision",

    hue="Feature"

)

plt.title("Precision Comparison")

plt.ylim(0,1)

plt.show()
```

# 7. RECALL

```python
plt.figure(figsize=(10,5))

sns.barplot(

    data=comparison,

    x="Model",

    y="Recall",

    hue="Feature"

)

plt.title("Recall Comparison")

plt.ylim(0,1)

plt.show()
```

# 8. F1-SCORE

```python
plt.figure(figsize=(10,5))

sns.barplot(

    data=comparison,

    x="Model",

    y="F1 Score",

    hue="Feature"

)

plt.title("Macro F1 Score Comparison")

plt.ylim(0,1)

plt.show()
```

# 9. BEST MODEL

```python
best_model = comparison.sort_values(

    by="F1 Score",

    ascending=False

).iloc[0]

best_model

print(f"""
Model    : {best_model['Model']}

Feature  : {best_model['Feature']}

Accuracy : {best_model['Accuracy']:.4f}

Precision: {best_model['Precision']:.4f}

Recall   : {best_model['Recall']:.4f}

F1 Score : {best_model['F1 Score']:.4f}
""")
```
